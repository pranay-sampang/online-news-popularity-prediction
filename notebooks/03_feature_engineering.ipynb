# Feature Engineering

### Import the required libraries

```python
import numpy as np
import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

### Load the data

```python
df = pd.read_csv('../data/OnlineNewsPopularityCleaned.csv')

df.head()
```

### Drop redundant/multicollinear features

```python
drop_cols = [
    'url',                        # web address label, no predictive signal
    'timedelta',                  # data collection artifact, not article quality
    'n_non_stop_words',           # near-constant value ≈ 1.0, zero variance
    'rate_negative_words',        # corr = -0.997 with rate_positive_words
    'kw_max_min',                 # corr = 0.940 with kw_avg_min
    'kw_max_avg',                 # corr = 0.816 with kw_avg_avg
    'n_non_stop_unique_tokens',   # corr = 0.885 with n_unique_tokens
    'self_reference_min_shares',  # redundant with self_reference_avg_shares
    'self_reference_max_shares',  # redundant with self_reference_avg_shares
]

df.drop(columns=drop_cols, inplace=True)

print(f"Dropped {len(drop_cols)} redundant features. Shape: {df.shape}")
```

### Log-transform skewed features (Using log1p (log(x+1)) to safely handle zeros)

```python
log_cols = [
    'n_tokens_content',
    'num_hrefs',
    'num_self_hrefs',
    'num_imgs',
    'num_videos',
    'kw_avg_min',
    'kw_min_max',
    'self_reference_avg_shares',
    'kw_avg_avg',
    'kw_avg_max',
    'shares',
]

for col in log_cols:
    df[f'{col}_log'] = np.log1p(df[col])
    df.drop(columns=[col], inplace=True)

print(f"Log-transformed {len(log_cols)} skewed features.")
```

```python
df.head()
```

### Engineer new features

```python
# From EDA: very short (<231) or very long (>824) articles perform better
df['is_short_article']  = (df['n_tokens_content_log'] < np.log1p(231)).astype(int)
df['is_long_article']   = (df['n_tokens_content_log'] > np.log1p(824)).astype(int)

# From EDA: images and videos linked to higher shares
df['has_image'] = (df['num_imgs_log'] > 0).astype(int)
df['has_video'] = (df['num_videos_log'] > 0).astype(int)

# Content richness: words per link (proxy for content quality)
df['content_to_links_ratio'] = df['n_tokens_content_log'] / (df['num_hrefs_log'] + 1)

# Keyword quality: interaction of avg keyword popularity with count
df['keyword_quality_score'] = df['kw_avg_avg_log'] * df['num_keywords']

# Sentiment polarity gap between title and body
df['sentiment_gap'] = abs(df['title_sentiment_polarity'] - df['global_sentiment_polarity'])

# is_weekend already exists in the data — confirm it's there
if 'is_weekend' not in df.columns:
    day_cols = ['weekday_is_saturday', 'weekday_is_sunday']
    df['is_weekend'] = df[day_cols].max(axis=1)

print(f"Engineered 7 new features. Shape: {df.shape}")
```

#### Final sanity check

```python
print(f"\nFinal dataset shape: {df.shape}")
print(f"\nNo nulls: {df.isnull().sum().sum() == 0}")
print(df.describe().round(3))
```

#### Save the cleaned data and Outputs

```python
df.to_csv('../data/OnlineNewsPopularityEngineered.csv', index=False)

print("\nSaved: OnlineNewsPopularityEngineered.csv")
```
