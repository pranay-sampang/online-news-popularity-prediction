# Data Processing

### Import the required libraries

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

### Load the data

```python
df = pd.read_csv('../data/OnlineNewsPopularity.csv')

df.head()
```

#### Dataset Shape

```python
dataset_shape = df.shape

print(f"Shape of dataset: {dataset_shape} | No. of Rows: {dataset_shape[0]:,} | No. of Columns: {dataset_shape[1]}")
```

### Check for missing values

```python
missing_values = df.isnull().sum().sum()

print(f"\nTotal missing values: {missing_values}")
```

### Check for duplicate rows and remove (if any)

```python
duplicate_rows = df.duplicated().sum()
print(f"\nNumber of duplicate rows before removal: {duplicate_rows}")

if duplicate_rows > 0:
    df = df.drop_duplicates()
    print(f"Removed {duplicate_rows} duplicate rows.")
    print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")
else:
    print("No duplicate rows found.")
```

### Strip leading and trailing spaces from every columns

Every column except 'url' has a leading space, e.g. ' timedelta', ' n_tokens_title', ' shares'. 
This breaks any dataframe column access.


```python
print("\nBefore: First 5 column names:")

for col in df.columns[:5]:
    print(f"{repr(col)}")
 
df.columns = df.columns.str.strip()
 
print("\nAfter: First 5 column names:")

for col in df.columns[:5]:
    print(f"{repr(col)}")
```

#### Fix column name typo

'self_reference_avg_sharess' has a double 's' at the end.

```python
typo_columns = [c for c in df.columns if 'avg_share' in c]

print(f"\nBefore: {typo_columns}")
 
df = df.rename(columns={'self_reference_avg_sharess': 'self_reference_avg_shares'})
 
print(f"\nAfter: {[c for c in df.columns if 'avg_share' in c]}")
```

#### Remove articles with zero content tokens 

1,181 rows have n_tokens_content == 0, the articles with no body.
These are likely scraping errors or stub pages. 
Removed rather than imputed since there is no meaningful content to recover.

```python
empty_tokens_content = (df['n_tokens_content'] == 0).sum()

print(f"\nBefore: Number of rows of articles with no body - {empty_tokens_content}")

df = df[df['n_tokens_content'] > 0]

print(f"\nAfter: Number of rows of articles with no body - {(df['n_tokens_content'] == 0).sum()}, {len(df)} rows remaining")
```

#### Replace keyword sentinel values (-1) with NaN, then impute with median

Articles with no keyword data have kw_* columns set to -1 as a sentinel value.
Since -1 is not meaningful, it is first replaced with NaN and then imputed using
the median of each column. Median is preferred over mean due to robustness against skewness.



```python
kw_columns = [
    'kw_min_min', 'kw_max_min', 'kw_avg_min',
    'kw_min_max', 'kw_max_max', 'kw_avg_max',
    'kw_min_avg', 'kw_max_avg', 'kw_avg_avg'
]

n_sentinel = (df[kw_columns] == -1).sum().sum()

print(f"\nBefore: Total sentinel (-1) values: {n_sentinel}")

df[kw_columns] = df[kw_columns].replace(-1, np.nan)
df[kw_columns] = df[kw_columns].fillna(df[kw_columns].median())

print(f"\nAfter: All sentinel values replaced and imputed using median")
```

#### Clip invalid values in Ratio-based features to valid range [0, 1]

Ratio-based features represent proportions (e.g., unique words / total words)
and must lie within the valid range [0, 1]. Values greater than 1 indicate
data inconsistencies or calculation errors. Since only a small number of rows
are affected, these values are clipped to 1.0 instead of removing the rows



```python
ratio_columns = [
    'n_unique_tokens',
    'n_non_stop_words',
    'n_non_stop_unique_tokens'
]

print("\nBefore: Invalid values (>1):")
for col in ratio_columns:
    print(f"{col}: {(df[col] > 1).sum()}")

df[ratio_columns] = df[ratio_columns].clip(upper=1.0)

print("\nAfter: Values clipped to maximum of 1.0")
```

#### Final sanity check

```python
print(f"\nFinal shape: {df.shape}")
print("\nSummary statistics:\n")
print(df.describe())
```

#### Save the cleaned data

```python
df.to_csv('../data/OnlineNewsPopularityCleaned.csv', index=False)

print(f"Cleaned data saved. Shape: {df.shape}")
```
